```python
# basic sereis fundamental 
s = pd.Series([1,2,3,4,5])
print(s.dtype)   # data type of series
print(s.size)    # size of series
print(s.ndim)    # dimension of series
print(s.shape)   # shape of series
print(s.empty)   # check if series is empty or not
print(s.hasnans) # check if series has null values or not
print(s.is_unique) # check if all values in series are unique or not
print(s.nunique()) # number of unique values in series  
print(s.values)   # values of series in array form
print(s.index)    # index of series
print(s.head(3))  # first 3 elements of series
print(s.tail(3))  # last 3 elements of series
print(s.describe()) # statistical summary of series
print(s.mean())    # mean of series
print(s.median())  # median of series
print(s.mode())    # mode of series
print(s.min())     # minimum value of series
print(s.max())     # maximum value of series
print(s.sum())     # sum of series
print(s.std())     # standard deviation of series
print(s.var())     # variance of series
print(s.cumsum())  # cumulative sum of series
print(s.cumprod()) # cumulative product of series   
print(s.value_counts()) # count of unique values in series
print(s.sort_values()) # sort series values
print(s.sort_index())  # sort series index
print(s.unique())     # unique values in series
print(s.sample(3))   # random sample of 3 elements from series
print(s.clip(2,4))  # clip values in series between 2 and
print(s.where(s>2)) # replace values in series less than 2 with NaN
print(s.replace(2,20)) # replace all occurrences of 2 with 20
print(s.shift(2))   # shift series values by 2 positions
print(s.diff())     # difference between consecutive elements in series
print(s.pct_change()) # percentage change between consecutive elements in series
print(s.corr(s))    # correlation between series and itself
print(s.cov(s))     # covariance between series and itself
print(s.isin([2,3])) # check if values in series are in the list
print(s.duplicated()) # check for duplicate values in series
print(s.drop_duplicates()) # drop duplicate values in series    # drop duplicate values in series
print(s.astype(float)) # convert series to float type
print(s.tolist())   # convert series to list
print(s.to_dict())  # convert series to dictionary
print(s.to_frame()) # convert series to dataframe
print(s.to_json())  # convert series to json
print(s.to_csv('series.csv')) # save series to csv file
print(s.to_excel('series.xlsx')) # save series to excel file
print(s.plot())     # plot series
print(s.hist())    # plot histogram of series
print(s.boxplot()) # plot boxplot of series
print(s.plot(kind='bar')) # plot bar chart of series
print(s.plot(kind='line')) # plot line chart of series
print(s.plot(kind='scatter', x=s.index, y=s.values)) # plot scatter plot
print(s.plot(kind='area')) # plot area chart of series
print(s.plot(kind='pie'))  # plot pie chart of series
print(s.rolling(window=2).mean()) # rolling mean with window size 2
print(s.rolling(window=2).sum())  # rolling sum with window size 2
print(s.expanding().mean()) # expanding mean
print(s.expanding().sum())  # expanding sum
print(s.ewm(span=2).mean()) # exponentially weighted mean with span 2
print(s.ewm(span=2).sum())  # exponentially weighted sum with span 2
print(s.pipe(lambda x: x*2)) # apply function to series using pipe
print(s.aggregate(['mean','sum','min','max'])) # aggregate functions on series
print(s.apply(lambda x: x*2)) # apply function to each element in series
print(s.map(lambda x: x*2))   # map function to each element in series
print(s.filter(like='2')) # filter series values containing '2'
print(s.where(s>2, other=0)) # replace values in series less than 2 with 0
print(s.mask(s>2, other=0))  # replace values in series greater than 2 with 0
print(s.between(2,4)) # check if values in series are between 2 and 4
print(s.clip(lower=2, upper=4)) # clip values in series between 2 and 4
print(s.nlargest(3)) # get 3 largest values in series
print(s.nsmallest(3)) # get 3 smallest values in series
print(s.sample(frac=0.5)) # random sample of 50% elements from series
print(s.sample(n=3))   # random sample of 3 elements from series
print(s.memory_usage()) # memory usage of series
```
